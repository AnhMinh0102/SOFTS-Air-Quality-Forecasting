```python
%%writefile models/Transformer.py

import torch
import torch.nn as nn
from layers.Embed import DataEmbedding_inverted

class Model(nn.Module):

    def __init__(self, configs):
        super().__init__()

        self.pred_len = configs.pred_len

        self.enc_embedding = DataEmbedding_inverted(
            configs.seq_len,
            configs.d_model,
            configs.dropout
        )

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=configs.d_model,
            nhead=8,
            dim_feedforward=configs.d_ff,
            dropout=configs.dropout,
            batch_first=True
        )

        self.encoder = nn.TransformerEncoder(
            encoder_layer,
            num_layers=configs.e_layers
        )

        self.projector = nn.Linear(
            configs.d_model,
            configs.pred_len
        )

    def forward(
        self,
        x_enc,
        x_mark_enc,
        x_dec,
        x_mark_dec,
        mask=None
    ):

        enc_out = self.enc_embedding(
            x_enc,
            None
        )

        enc_out = self.encoder(enc_out)

        enc_out = self.projector(enc_out)

        enc_out = enc_out.permute(0, 2, 1)

        return enc_out
```

```python
import torch
import matplotlib.pyplot as plt
import numpy as np
from data_provider.data_factory import data_provider
from models import SOFTS, Transformer

# =====================================================
# 1. Khởi tạo tham số
# =====================================================

class DummyArgs:
    pass

args = DummyArgs()

args.task_name = 'long_term_forecast'
args.data = 'ETTm1'
args.root_path = './dataset/'
args.data_path = 'ETTm1.csv'

args.features = 'M'
args.target = 'OT'
args.freq = 'h'

args.seq_len = 96
args.label_len = 48
args.pred_len = 96

args.enc_in = 7
args.dec_in = 7
args.c_out = 7

args.d_model = 128
args.d_core = 512
args.e_layers = 2
args.d_layers = 1
args.d_ff = 512

args.dropout = 0.0
args.embed = 'timeF'
args.activation = 'gelu'

args.output_attention = False
args.use_norm = True

args.batch_size = 32
args.num_workers = 2
args.seasonal_patterns = 'Monthly'

device = torch.device(
    'cuda:0' if torch.cuda.is_available() else 'cpu'
)

# =====================================================
# 2. Load dữ liệu Test
# =====================================================

dataset, dataloader = data_provider(args, 'test')

data_iter = iter(dataloader)

batch_x, batch_y, batch_x_mark, batch_y_mark = next(data_iter)

batch_x = batch_x.float().to(device)
batch_y = batch_y.float().to(device)

batch_x_mark = batch_x_mark.float().to(device)
batch_y_mark = batch_y_mark.float().to(device)

# decoder input
dec_inp = torch.zeros_like(
    batch_y[:, -args.pred_len:, :]
).float()

dec_inp = torch.cat(
    [batch_y[:, :args.label_len, :], dec_inp],
    dim=1
).float().to(device)

# =====================================================
# 3. Load SOFTS
# =====================================================

model_softs = SOFTS.Model(args).to(device)

softs_path = "./checkpoints/long_term_forecast_SOFTS_ETTm1_SOFTS_ETTm1_ftM_sl96_ll48_pl96_dm128_el2_dl1_df512_fc1_ebtimeF_dtTrue_test/checkpoint.pth"

model_softs.load_state_dict(
    torch.load(softs_path, map_location=device)
)

model_softs.eval()

# =====================================================
# 4. Load Transformer
# =====================================================

model_trans = Transformer.Model(args).to(device)

transformer_path = "./checkpoints/long_term_forecast_Transformer_ETTm1_Transformer_ETTm1_ftM_sl96_ll48_pl96_dm128_el2_dl1_df512_fc1_ebtimeF_dtTrue_Transformer_Baseline/checkpoint.pth"

model_trans.load_state_dict(
    torch.load(transformer_path, map_location=device)
)

model_trans.eval()

# =====================================================
# 5. Inference
# =====================================================

with torch.no_grad():

    out_softs = model_softs(
        batch_x,
        batch_x_mark,
        dec_inp,
        batch_y_mark
    )

    out_trans = model_trans(
        batch_x,
        batch_x_mark,
        dec_inp,
        batch_y_mark
    )

# =====================================================
# 6. Chọn mẫu để hiển thị
# =====================================================

sample_idx = 0

# OT (cột cuối)
target_var_idx = -1

pred_softs = out_softs[
    sample_idx,
    -args.pred_len:,
    target_var_idx
].cpu().numpy()

pred_trans = out_trans[
    sample_idx,
    -args.pred_len:,
    target_var_idx
].cpu().numpy()

true = batch_y[
    sample_idx,
    -args.pred_len:,
    target_var_idx
].cpu().numpy()

history = batch_x[
    sample_idx,
    :,
    target_var_idx
].cpu().numpy()

# =====================================================
# 7. Visualize
# =====================================================

plt.figure(figsize=(14,6), dpi=100)

x_history = np.arange(args.seq_len)

x_future = np.arange(
    args.seq_len,
    args.seq_len + args.pred_len
)

plt.plot(
    x_history,
    history,
    label='History (96 bước)',
    color='gray',
    linewidth=2
)

plt.plot(
    x_future,
    true,
    label='Ground Truth',
    color='black',
    linewidth=2.5
)

plt.plot(
    x_future,
    pred_softs,
    label='SOFTS',
    color='red',
    linestyle='--',
    linewidth=2
)

plt.plot(
    x_future,
    pred_trans,
    label='Transformer',
    color='blue',
    linestyle='-.',
    linewidth=2
)

plt.axvline(
    x=args.seq_len,
    color='green',
    linestyle=':',
    label='Forecast Start'
)

plt.title(
    'SOFTS vs Transformer on ETTm1 (OT Variable)',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Time Steps')
plt.ylabel('Normalized Value')

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import torch
import matplotlib.pyplot as plt
import numpy as np
import urllib.request
from matplotlib.font_manager import FontProperties
from data_provider.data_factory import data_provider
from models import SOFTS, Transformer

# ==========================================
# 1. FONT TIẾNG VIỆT
# ==========================================
font_url = 'https://github.com/googlefonts/roboto/raw/main/src/hinted/Roboto-Regular.ttf'
urllib.request.urlretrieve(font_url, 'Roboto-Regular.ttf')
vi_font = FontProperties(fname='Roboto-Regular.ttf')

# ==========================================
# 2. CONFIG
# ==========================================
class DummyArgs:
    pass

args = DummyArgs()
args.task_name = 'long_term_forecast'
args.data = 'custom'
args.root_path = './dataset/'
args.data_path = 'Cleaned_Hanoi_Data.csv'
args.features = 'M'
args.target = 'pm10'
args.freq = 'h'

args.seq_len = 96
args.label_len = 48
args.pred_len = 96

args.enc_in = 7
args.dec_in = 7
args.c_out = 7

args.d_model = 512
args.d_core = 512
args.e_layers = 2
args.d_layers = 1
args.d_ff = 2048
args.dropout = 0.0
args.embed = 'timeF'
args.activation = 'gelu'
args.output_attention = False
args.use_norm = True

args.batch_size = 32
args.num_workers = 2
args.seasonal_patterns = 'Monthly'

device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')

# ==========================================
# 3. DATA
# ==========================================
dataset, dataloader = data_provider(args, 'test')
batch_x, batch_y, batch_x_mark, batch_y_mark = next(iter(dataloader))

batch_x = batch_x.float().to(device)
batch_y = batch_y.float().to(device)
batch_x_mark = batch_x_mark.float().to(device)
batch_y_mark = batch_y_mark.float().to(device)

dec_inp = torch.zeros_like(batch_y[:, -args.pred_len:, :]).float()
dec_inp = torch.cat([batch_y[:, :args.label_len, :], dec_inp], dim=1).float().to(device)

# ==========================================
# 4. LOAD MODELS
# ==========================================

# ===== SOFTS =====
args.model = 'SOFTS'
model_softs = SOFTS.Model(args).to(device)

softs_path = './checkpoints/long_term_forecast_SOFTS_Hanoi_AirQuality_SOFTS_custom_ftM_sl96_ll48_pl96_dm512_el2_dl1_df2048_fc1_ebtimeF_dtTrue_test/checkpoint.pth'

ckpt_softs = torch.load(softs_path, map_location=device)
model_softs.load_state_dict(ckpt_softs)
model_softs.eval()

# ===== TRANSFORMER =====
args.model = 'Transformer'
model_transformer = Transformer.Model(args).to(device)

transformer_path = './checkpoints/long_term_forecast_Transformer_Hanoi_AirQuality_Transformer_custom_ftM_sl96_ll48_pl96_dm512_el2_dl1_df2048_fc1_ebtimeF_dtTrue_So_sanh_Baseline_Hanoi/checkpoint.pth'

ckpt_tr = torch.load(transformer_path, map_location=device)
model_transformer.load_state_dict(ckpt_tr)
model_transformer.eval()

# =====================================================
# 5. INFERENCE & TÍNH TOÁN CHỈ SỐ ĐÁNH GIÁ (METRICS)
# =====================================================
with torch.no_grad():
    # Dự báo trên toàn bộ dataloader (tập test) để có thống kê chuẩn xác nhất
    all_true = []
    all_pred_softs = []
    all_pred_tr = []

    for batch_x, batch_y, batch_x_mark, batch_y_mark in dataloader:
        batch_x = batch_x.float().to(device)
        batch_y = batch_y.float().to(device)
        batch_x_mark = batch_x_mark.float().to(device)
        batch_y_mark = batch_y_mark.float().to(device)

        dec_inp = torch.zeros_like(batch_y[:, -args.pred_len:, :]).float()
        dec_inp = torch.cat([batch_y[:, :args.label_len, :], dec_inp], dim=1).float().to(device)

        out_softs = model_softs(batch_x, batch_x_mark, dec_inp, batch_y_mark)
        out_tr = model_transformer(batch_x, batch_x_mark, dec_inp, batch_y_mark)

        # Chỉ lấy phần chuỗi tương lai cần dự báo (pred_len)
        all_true.append(batch_y[:, -args.pred_len:, :].cpu().numpy())
        all_pred_softs.append(out_softs[:, -args.pred_len:, :].cpu().numpy())
        all_pred_tr.append(out_tr[:, -args.pred_len:, :].cpu().numpy())

    # Gộp toàn bộ các batch lại thành tensor hoàn chỉnh
    all_true = np.concatenate(all_true, axis=0)
    all_pred_softs = np.concatenate(all_pred_softs, axis=0)
    all_pred_tr = np.concatenate(all_pred_tr, axis=0)

# Định nghĩa hàm tính MAE và MSE chuẩn
def calculate_metrics(true, pred):
    mse = np.mean((true - pred) ** 2)
    mae = np.mean(np.abs(true - pred))
    return mse, mae

# Định nghĩa danh sách các khí và index tương ứng (bỏ cột 'date')
targets = {
    'PM10': 6,
    'NO2': 1,
    'CO': 0
}

# Khởi tạo danh sách chứa kết quả thống kê
metrics_summary = []

for gas_name, var_idx in targets.items():
    # Lấy lát cắt dữ liệu thực tế và dự báo của riêng khí đó trên toàn bộ tập test
    true_gas = all_true[:, :, var_idx]
    softs_gas = all_pred_softs[:, :, var_idx]
    tr_gas = all_pred_tr[:, :, var_idx]

    # Tính toán lỗi cho mô hình SOFTS
    mse_s, mae_s = calculate_metrics(true_gas, softs_gas)

    # Tính toán lỗi cho mô hình Transformer
    mse_tr, mae_tr = calculate_metrics(true_gas, tr_gas)

    # Lưu kết quả vào danh sách
    metrics_summary.append({
        'Chất khí': gas_name,
        'SOFTS_MSE': mse_s,
        'SOFTS_MAE': mae_s,
        'Transformer_MSE': mse_tr,
        'Transformer_MAE': mae_tr
    })

# =====================================================
# 6. HIỂN THỊ BẢNG THỐNG KÊ CHỈ SỐ ĐÁNH GIÁ
# =====================================================
import pandas as pd

# Chuyển đổi sang DataFrame để hiển thị đẹp mắt
df_metrics = pd.DataFrame(metrics_summary)

print("\n" + "="*33 + " BẢNG THỐNG KÊ CHỈ SỐ ĐÁNH GIÁ TRÊN TẬP TEST (96h) " + "="*33)
display(df_metrics.round(5)) # Làm tròn 5 chữ số thập phân cho gọn sạch
print("="*118)

# Vẽ biểu đồ cột (Bar plot) so sánh MAE giữa 2 mô hình cho cả 3 khí
import seaborn as sns

# Biến đổi cấu trúc DataFrame một chút để phục vụ vẽ đồ thị nhóm
plot_data = []
for m in metrics_summary:
    plot_data.append({'Gas': m['Chất khí'], 'Model': 'SOFTS', 'MSE': m['SOFTS_MSE'], 'MAE': m['SOFTS_MAE']})
    plot_data.append({'Gas': m['Chất khí'], 'Model': 'Transformer', 'MSE': m['Transformer_MSE'], 'MAE': m['Transformer_MAE']})
df_plot = pd.DataFrame(plot_data)

fig, axes = plt.subplots(1, 2, figsize=(14, 5), dpi=100)

# Vẽ biểu đồ MSE
sns.barplot(x='Gas', y='MSE', hue='Model', data=df_plot, ax=axes[0], palette=['#ff4d4d', '#4d79ff'])
axes[0].set_title('So sánh sai số MSE (Thấp hơn là tốt hơn)', fontproperties=vi_font, fontsize=12, fontweight='bold')
axes[0].set_ylabel('Mean Squared Error', fontproperties=vi_font)
axes[0].set_xlabel('Loại khí', fontproperties=vi_font)
axes[0].legend(prop=vi_font)

# Vẽ biểu đồ MAE
sns.barplot(x='Gas', y='MAE', hue='Model', data=df_plot, ax=axes[1], palette=['#ff4d4d', '#4d79ff'])
axes[1].set_title('So sánh sai số MAE (Thấp hơn là tốt hơn)', fontproperties=vi_font, fontsize=12, fontweight='bold')
axes[1].set_ylabel('Mean Absolute Error', fontproperties=vi_font)
axes[1].set_xlabel('Loại khí', fontproperties=vi_font)
axes[1].legend(prop=vi_font)

plt.tight_layout()
plt.show()
```

```python
# =====================================================
# 1. INFERENCE TRÊN TOÀN BỘ TẬP TEST
# =====================================================
with torch.no_grad():
    all_true = []
    all_pred_softs = []
    all_pred_tr = []

    for batch_x, batch_y, batch_x_mark, batch_y_mark in dataloader:
        batch_x = batch_x.float().to(device)
        batch_y = batch_y.float().to(device)
        batch_x_mark = batch_x_mark.float().to(device)
        batch_y_mark = batch_y_mark.float().to(device)

        dec_inp = torch.zeros_like(batch_y[:, -args.pred_len:, :]).float()
        dec_inp = torch.cat([batch_y[:, :args.label_len, :], dec_inp], dim=1).float().to(device)

        out_softs = model_softs(batch_x, batch_x_mark, dec_inp, batch_y_mark)
        out_tr = model_transformer(batch_x, batch_x_mark, dec_inp, batch_y_mark)

        all_true.append(batch_y[:, -args.pred_len:, :].cpu().numpy())
        all_pred_softs.append(out_softs[:, -args.pred_len:, :].cpu().numpy())
        all_pred_tr.append(out_tr[:, -args.pred_len:, :].cpu().numpy())

    all_true = np.concatenate(all_true, axis=0)
    all_pred_softs = np.concatenate(all_pred_softs, axis=0)
    all_pred_tr = np.concatenate(all_pred_tr, axis=0)

# =====================================================
# 2. NGHỊCH ĐẢO CHUẨN HÓA CHUẨN XÁC CHO TỪNG CỘT
# =====================================================
scaler = dataset.scaler

# Khởi tạo các mảng chứa dữ liệu đã giải chuẩn hóa
all_true_scaled = np.zeros_like(all_true)
all_pred_softs_scaled = np.zeros_like(all_pred_softs)
all_pred_tr_scaled = np.zeros_like(all_pred_tr)

# Giải chuẩn hóa cho dữ liệu tương lai
for t in range(args.pred_len):
    all_true_scaled[:, t, :] = scaler.inverse_transform(all_true[:, t, :])
    all_pred_softs_scaled[:, t, :] = scaler.inverse_transform(all_pred_softs[:, t, :])
    all_pred_tr_scaled[:, t, :] = scaler.inverse_transform(all_pred_tr[:, t, :])

# Trích xuất và giải chuẩn hóa cho dữ liệu lịch sử
batch_x_raw, _, _, _ = next(iter(dataloader))
batch_x_raw = batch_x_raw.numpy()
history_scaled_batch = np.zeros_like(batch_x_raw)

for t in range(args.seq_len):
    history_scaled_batch[:, t, :] = scaler.inverse_transform(batch_x_raw[:, t, :])

# =====================================================
# CẤU HÌNH INDEX THỰC TẾ SAU KHI TARGET BỊ ĐẨY XUỐNG CUỐI
# =====================================================
# Giải thích: CSV gốc gồm [pm10, co, no2, temp, humid, precip, wind] (7 cột không tính date)
# Khi set target='pm10', pm10 nhảy xuống cuối thành: [co, no2, temp, humid, precip, wind, pm10]
targets = {
    'CO': 0,
    'NO2': 1,
    'PM10': 6
}

# =====================================================
# 3. TÍNH CHỈ SỐ ĐÁNH GIÁ TRÊN THANG ĐO THỰC TẾ
# =====================================================
def calculate_metrics(true, pred):
    mse = np.mean((true - pred) ** 2)
    mae = np.mean(np.abs(true - pred))
    return mse, mae

metrics_summary = []
for gas_name, var_idx in targets.items():
    true_gas = all_true_scaled[:, :, var_idx]
    softs_gas = all_pred_softs_scaled[:, :, var_idx]
    tr_gas = all_pred_tr_scaled[:, :, var_idx]

    mse_s, mae_s = calculate_metrics(true_gas, softs_gas)
    mse_tr, mae_tr = calculate_metrics(true_gas, tr_gas)

    metrics_summary.append({
        'Chất khí': gas_name,
        'SOFTS_MSE': mse_s,
        'SOFTS_MAE': mae_s,
        'Transformer_MSE': mse_tr,
        'Transformer_MAE': mae_tr
    })

import pandas as pd
df_metrics = pd.DataFrame(metrics_summary)
print("\n" + "="*28 + " BẢNG CHỈ SỐ ĐÁNH GIÁ TRÊN THANG ĐO GỐC " + "="*28)
display(df_metrics.round(3))
print("="*109)

# =====================================================
# 4. VẼ BIỂU ĐỒ ĐƯỜNG (SUBPLOTS) THEO THANG ĐO THỰC TẾ (24 GIỜ)
# =====================================================
sample_idx = 0
plot_len = 24

fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(14, 16), dpi=100, sharex=True)

x_history = np.arange(args.seq_len)
x_future = np.arange(args.seq_len, args.seq_len + plot_len)

# Sử dụng danh sách sắp xếp lại để vẽ theo đúng thứ tự hiển thị mong muốn (PM10 -> CO -> NO2)
order_to_plot = ['PM10', 'CO', 'NO2']

for idx, gas_name in enumerate(order_to_plot):
    var_idx = targets[gas_name]
    ax = axes[idx]

    # Lấy dữ liệu theo đúng index thực tế của scaler
    history = history_scaled_batch[sample_idx, :, var_idx]
    true_ = all_true_scaled[sample_idx, :plot_len, var_idx]
    pred_s = all_pred_softs_scaled[sample_idx, :plot_len, var_idx]
    pred_tr = all_pred_tr_scaled[sample_idx, :plot_len, var_idx]

    # Vẽ đồ thị
    ax.plot(x_history, history, label=f'Lịch sử {gas_name} (96 giờ)', color='gray', linewidth=2)
    ax.plot(x_future, true_, label=f'Thực tế {gas_name} (24 giờ)', color='black', linewidth=2.5)
    ax.plot(x_future, pred_s, label=f'SOFTS Dự báo {gas_name}', color='red', linestyle='--', linewidth=2)
    ax.plot(x_future, pred_tr, label=f'Transformer Dự báo {gas_name}', color='blue', linestyle='-.', linewidth=2)

    ax.axvline(x=args.seq_len, color='green', linestyle=':', label='Hiện tại')

    ax.set_title(f'Dự báo khí {gas_name} trên thang đo thực tế (SOFTS vs Transformer)', fontproperties=vi_font, fontsize=14, fontweight='bold')
    ax.set_ylabel(f'Nồng độ {gas_name}', fontproperties=vi_font, fontsize=12)
    ax.legend(prop=vi_font, loc='upper left')
    ax.grid(True, alpha=0.3)

axes[-1].set_xlabel('Thời gian (Giờ)', fontproperties=vi_font, fontsize=12)
plt.tight_layout()
plt.show()
```
